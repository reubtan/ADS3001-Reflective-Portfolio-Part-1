###**How-To Guide: Event-Driven Logic with Dash Callbacks**

This notebook demonstrates how I learnt to use event-driven logic and callback architecture in Plotly Dash this week.

A key difference between a traditional Python script and an interactive dashboard is that a dashboard needs to respond to actions taken by the user. Dash callbacks allow the application to detect changes, execute Python functions and update the interface automatically.

This guide introduces the basic concept before demonstrating a simple working example.

##**What I learnt**

Since this week my team was ahead of schedule and beginning the preparation phase, I began learning how interactive dashboards are structured using Plotly Dash. One of the concepts I found important was event-driven logic.

Before working with Dash, I was more familiar with Python programs that followed a sequential process. The code would run from the beginning to the end, process some data and produce an output. Dash required me to think differently because the application needed to wait for actions from the user.

For example, a user might enter a search term, click a button or change a value. Instead of manually running another section of Python code, the dashboard can detect the event and automatically execute the appropriate callback.

Learning this changed how I approached application development. I started thinking about what should trigger an action, what information the function needed and which part of the interface should change as a result.

##**What is event-driven logic?**

Event-driven logic means that a program responds to specific events rather than simply executing every instruction sequentially.

In a Dash application, the general process is:

User action → Input changes → Callback runs → Output changes

For example:

User clicks button → Button's n_clicks changes → Dash detects the change → Callback function executes → Dashboard output updates


```python
!pip install dash jupyter-dash
```

```python
from dash import Dash, html, Input, Output

app = Dash(__name__)

app.layout = html.Div([
    html.H2("Simple Dash Callback"),

    html.Button(
        "Click me",
        id="button",
        n_clicks=0
    ),

    html.Div(
        id="output",
        style={"marginTop": "20px", "fontSize": "20px"}
    )
])


@app.callback(
    Output("output", "children"),
    Input("button", "n_clicks")
)
def update_output(n_clicks):
    return f"The button has been clicked {n_clicks} times."
```

##**Breaking down the callback**

The important part of the example above is:



```python
@app.callback(
    Output("output", "children"),
    Input("button", "n_clicks")
  )
```

##**Why this is different from a normal Python script**

The callback does not continuously run in the background.

Instead, Dash creates a relationship between the components:

button.n(clicks) → update_output() → output.children


##**Connecting the concept to my internship project**

This simple example demonstrates the same underlying idea that I implemented in our testing for the Keystone Employer Database during the week.

The test dashboard contains a few filters like the one below:



```python
from jupyter_dash import JupyterDash #Using JupterDash for the sake of showcasing the filter
from dash import html, dcc, Input, Output
import pandas as pd


businesses = pd.DataFrame({ #Example Business Data
    "Business Name": [
        "ABC Café",
        "ABC Bakery",
        "ABC Gym",
        "ABC Grocer",
        "ABC Book Shop",
        "ABC Studio"
    ],
    "Industry": [
        "Hospitality",
        "Hospitality",
        "Health & Fitness",
        "Retail",
        "Retail",
        "Health & Fitness"
    ]
})

app = JupyterDash(__name__) #Dash app

app.layout = html.Div([ #App layout
    html.H2("Business Filter"),

    dcc.Dropdown(
        id="industry-filter",
        options=[
            {"label": industry, "value": industry}
            for industry in sorted(businesses["Industry"].unique())
        ],
        placeholder="Select an industry"
    ),

    html.Div(
        id="business-results",
        style={"marginTop": "20px"}
    )
])

@app.callback(
    Output("business-results", "children"),
    Input("industry-filter", "value")
)
def filter_businesses(selected_industry):

    if not selected_industry:
        filtered = businesses
    else:
        filtered = businesses[
            businesses["Industry"] == selected_industry
        ]

    return [
        html.Div(
            f"{row['Business Name']} — {row['Industry']}",
            style={
                "padding": "10px",
                "borderBottom": "1px solid #ddd"
            }
        )
        for _, row in filtered.iterrows()
    ]

app.run(mode="inline", debug=False)
```

##**What I found challenging**

The main challenge when I first encountered callbacks when researching Dash was understanding that the function was not simply being called like a normal Python function. Instead, Dash controlled when the function executed based on changes to its Inputs.
I initially had to think carefully about the relationship between the interface and the Python code. A component needed an ID, its relevant property had to be identified and that property then needed to be connected to a callback.

As I became more comfortable with this architecture, I found it easier to break dashboard features into smaller interactions. Rather than thinking about the entire application at once, I could ask:

1. What action is the user taking?
2. Which component detects that action?
3. What information does the callback need?
4. What should change after the callback runs?

This way of thinking became useful as the dashboard developed because more complex features could be built by combining these smaller event-driven relationships.

##**Key takeaways**

The main technical lessons I took from learning Dash callbacks were:

- **Inputs define events** – they determine what causes a callback to run.
- **Outputs define changes** – they determine which part of the interface is updated.
- **Callbacks connect the UI to Python logic** – allowing the dashboard to respond to users.
- **State matters** – interactive applications need to keep track of information between events.
- **Complex applications can be built from simple interactions** – the same callback principles can be extended to larger dashboard features which we will use in the future to create our deliverable dashboard.
